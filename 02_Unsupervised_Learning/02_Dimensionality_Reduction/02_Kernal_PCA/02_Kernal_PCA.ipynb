```python
import numpy as np
```

```python
from sklearn.datasets import make_circles
```

```python
X, Y = make_circles(n_samples=400, factor=0.3, noise=0.05)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(X[:,0], X[:,1], c=Y)
plt.title("Original 2D Circle Data")
plt.show()
```

```python
from sklearn.decomposition import KernelPCA
```

```python
kernalpca = KernelPCA(n_components=1, kernel='rbf', gamma=15)
```

```python
X_kpca = kernalpca.fit_transform(X)
```

```python
plt.scatter(X_kpca, np.zeros_like(X_kpca), c=Y)
plt.title("Data after Kernel PCA (1D)")
plt.show()
```
