```python
import numpy as np
X = np.array([
    [1, 2], [1.5, 1.8], [2, 2],
    [8, 8], [8.5, 8.2], [9, 9]
])
```

```python
from sklearn.cluster import MeanShift

ms = MeanShift(bandwidth=2)
ms.fit(X)

print("Cluster Centers:", ms.cluster_centers_)
print("Labels:", ms.labels_)
```
