```python
import pandas as pd
```

```python
from sklearn.datasets import load_diabetes
dataset = load_diabetes()
```

```python
X = dataset.data
Y = dataset.target
```

Linear Regression

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, train_size=0.7, random_state=123)

from sklearn.linear_model import LinearRegression
LinearRegressionModel = LinearRegression()

LinearRegressionModel.fit(X_train, Y_train)

Y_pred = LinearRegressionModel.predict(X_test)

from sklearn.metrics import mean_squared_error
print(mean_squared_error(Y_pred, Y_test))

LinearRegressionModel.coef_
```

Ridge Regression

```python
from sklearn.linear_model import Ridge

alpha = 1.5
LinearRegressionRidgeModel = Ridge(alpha=alpha)

LinearRegressionRidgeModel.fit(X_train, Y_train)

YR_Pred = LinearRegressionRidgeModel.predict(X_test)

print(mean_squared_error(YR_Pred, Y_test))

LinearRegressionRidgeModel.coef_
```

LASSO Regression

```python
from sklearn.linear_model import Lasso

alpha = 0.1
LinearRegressionLassoModel = Lasso(alpha=alpha)

LinearRegressionLassoModel.fit(X_train, Y_train)

YL_pred = LinearRegressionLassoModel.predict(X_test)

print(mean_squared_error(YL_pred, Y_test))

LinearRegressionLassoModel.coef_
```
